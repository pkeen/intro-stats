# Paired T-test

## 1. Compute differences
$$d_i = x_{i1} - x_{i2}$$

```python
import numpy as np

# Lets pretend I already computed differences 
d = np.array([1.2, 2.4, 1.3, 1.3, 0, 1, 1.8, 0.8, 4.6, 1.4])
```

## 2. Compute mean of differences
$$ \bar d = \frac{1}{n} \sum d_i $$

```python
d_bar = np.mean(d)
```

## 3. Compute standard deviation of differences
$$ s_d = \sqrt{\frac{\sum(d_i - \bar{d})^2}{n-1}}$$

```python
# 3. Std dev of differences
s_d = np.std(d, ddof=1)   # IMPORTANT
## ddof - delta degrees of freedom - setting to 1 means the sample variance is used (n-1 in the denominator)
```

## 4. Compute test statistic
$$ t = \frac{\bar d - \mu_0}{s_d / \sqrt{n}} $$

Usually $\mu_0 = 0$

```python
# Sample size
n = len(d)

# 5. t statistic
t_stat = d_bar / (s_d / np.sqrt(n))

print(f"t-statistic: {t_stat:.4f}")
```

## 5. Degrees of freedom

$$ df = n - 1 $$

```python
df = len(d) - 1
```

## 6. Compare against critical value
Two tailed: 
$$ t_{crit} = t_{1-\alpha/2, df}$$

Decision rule: 
Reject if: 
$$ |t| > t_{crit} $$

```python
from scipy.stats import t

# 6. Critical value
alpha = 0.05
df = n - 1
t_crit = t.ppf(1 - alpha/2, df)
print(f"Critical value: {t_crit:.4f}")
```

```python
# Decision
if abs(t_stat) >= t_crit:
    print("Reject H0")
else:
    print("Fail to reject H0")

print(f"t_stat = {t_stat:.3f}, t_crit = ±{t_crit:.3f}")
```

## 7. BONUS - calculate p-value
The p-value answers:

“If the null hypothesis were true, how likely is it to observe a t-statistic at least as extreme as mine?”

```python
# Compute p-value
p_value = 2 * (1 - t.cdf(abs(t_stat), df))

print(f"p-value: {p_value:.6f}")
```
