```python
import numpy as np

x = np.array([33.4, 30.9, 33.0, 32.6, 33.5, 32.2, 31.3, 32.1, 32.8, 31.8, 32.2, 32.7, 33.1, 30.9, 31.4])
y = np.array([100, 98, 100, 99, 100, 99, 95, 95, 99, 97, 100, 100, 100, 91, 100])

print(np.sum(x))
print(np.sum(y))
print(np.sum(x**2))
```

```python
r = (np.sum(x*y) - np.sum(x)*np.sum(y)/len(x)) / np.sqrt((np.sum(x**2) - np.sum(x)**2/len(x)) * (np.sum(y**2) - np.sum(y)**2/len(y)))
print(r)
```
