```python
import pandas as pd

df = pd.read_csv('../../data/birthrates-2015.txt', sep='\t')
```

```python
# Scatter plot
import matplotlib.pyplot as plt

plt.scatter(df['Year_since_1900'], df['Rate'])
```

```python

import scipy.stats as stats

slope, intercept, rvalue, _, _ = stats.linregress(df['Year_since_1900'], df['Rate'])

birth_rate_pred = intercept + slope * df['Year_since_1900']
print(intercept, slope, rvalue)
print(f"The equatin for the line is {intercept:.2f} + {slope:.2f} * Year_since_1900")
```

```python
plt.scatter(df['Year_since_1900'], df['Rate'], label="Original Data");
plt.plot(df['Year_since_1900'], intercept + slope*df['Year_since_1900'], 'r', label='fitted line');
plt.legend();
```

```python
print(f"The predicted rate for 1978 is {intercept + slope * (1978 - 1900):.2f}")
```

```python
print(f"The prediction for 2020 is {intercept + slope * (2020 -1900):.2f}")
```
