# House Prices Exercise
A scatterplot of house Price (in dollars) vs. house Size (in square feet) for houses sold recently in Saratoga, New York, shows a relationship that is straight, with only moderate scatter and no outliers. The correlation between house Price $(P)$ and house Size $(S)$ is 0.76. The least squares line is
$$
\hat {P} = 8400 + 88.67S
$$

1. You go to an open house and find that the house is 1 SD above the mean Size. How many SDs above the mean Price would you predict it to cost?

I would predict it would be less than 1 SD above the mean, in fact id predict it to be 1 SD multiplied by the $R$ value. 

$$
\hat Z_{price} = R \cdot Z_{size}
$$

2. You read an ad for a house that’s 2 SDs below the mean Size. What would you guess about its price?

I would guess its SD value to be $$
\hat Z_{price} = R \cdot Z_{size} =  -2R
$$

3. A friend tells you about a house whose Size in square meters (he’s European) is 1.5 stan-dard deviations above the mean. What would you guess about its Size in square feet?

The size in square meters and size in square feet are different units measuring the same variable. As such they will have a $R$ value of 1. So its value in square feet will be 1.5 SD above the mean.
$$
\frac{Z_{size}\cdot m^2}{Z_{size}\cdot ft^2} = \frac{m}{ft}
$$

4. Suppose the standard deviation of the house Prices is $77,000. How much more expensive than the mean Price would you predict for the house in Part 1 (that’s one SD above the mean in Size)?

I would predict it to be 

$$ 
\Delta y = R \cdot S_y
$$
Subsituting the values:
$$
\Delta y = 0.76 \cdot 77{,}000 = 58{,}520
$$

5. Suppose the mean Price of all these houses is $168,000. What is the predicted Price for the house in Part 4 that’s one SD above the mean in Size?

$$
\hat y = \bar y + R \cdot S_y = 168{,}000 + 58{,}520 = 226{,}520
$$

The predicted price is $226,520

6. The mean Size of the houses is 1800 sq ft and the standard deviation is 660 sq ft. Predict the Price of a 2460 sq ft house (1 SD above the mean) using the least squares equation in the original units to verify that you get the same prediction as in Part 5.

first lets test given least squares line:
$$
    \hat P = 8400 + 88.67 \cdot size
$$

Lets use code:

```python
# Using given formula
price_predicted = 8400 + 88.67*(2460)
print(f"the predicted price is ${price_predicted}")
```

Now lets work it out with the formula for $\beta_1$

$$
\beta_1=r\frac{s_y}{s_x}
$$


```python
# Now using the supplied sd and mean values
x = 2460
x_bar = 1800
s_x = 660

r = 0.76
y_bar = 168_000
s_y = 77000

slope = r*(s_y/s_x)
intercept = y_bar - slope * x_bar

y_hat = intercept + slope * x
print(f"the predicted price is ${y_hat}")  # Should be 226520
```
