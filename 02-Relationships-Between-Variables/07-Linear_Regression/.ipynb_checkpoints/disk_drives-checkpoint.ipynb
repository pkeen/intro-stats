```python
import pandas as pd

df = pd.read_csv("../../data/disk_drives_2020.txt", sep="\t")
```

# Full Math
Im gong to to use numpy to calculate all the values for the linear regression

## Calculate the mean
$$
\bar{x} = \frac{1}{N} \sum_{i=1}^{N} x_i
$$


```python
import numpy as np

x = df["Capacity(TB)"].to_numpy()

x_bar = np.sum(x) / len(x)

y = df["Price($)"].to_numpy()

y_bar = np.sum(y) / len(y)
```

## Calculate Standard Deviation
$$
s_y = \sqrt{\frac{\Sigma(y_i - \bar{y})^2}{N-1}}
$$

```python
## Calculate Standard Deviation
s_x = np.sqrt(np.sum((x - x_bar)**2) / (len(x) - 1))
s_y = np.sqrt(np.sum((y - y_bar)**2) / (len(y) - 1))
```

## Calculate Z-Sxores and Correlation

Correlation
$$
r=\frac{\Sigma z_{x,i} \cdot z_{y,i}}{n-1}
$$

```python
# Assume x and y are 1D NumPy arrays of the same length
z_x = (x - np.mean(x)) / np.std(x)
z_y = (y - np.mean(y)) / np.std(y)

correlation = np.mean(z_x * z_y)
print(correlation)
```

## Find the Slope Estimate - $\Beta_1$
The formula for slope is:
$$
\beta_1=r\frac{s_y}{s_x}
$$

```python
b_1 = correlation * (s_y/s_x)
```

The slope is 41.729.

### What does the slope mean in this context?
It means that for every 1TB increase in storage capacity, we would expect to see a $41.73 increase in price. 

## Find the intercept - $\Beta_0$
$$
\beta_0 = \bar{y} - \beta_1 \bar{x}
$$

```python
# Calculate b0 (intercept)
b_0 = y_bar - b_1 * x_bar
print(b_0)
```

### What does the intercept mean in this context?
It is the hypoethetical value of y when x is 0.

## Write the equation predicting Price from Capacity
$$
\hat{y} = \beta_0 + \beta_1 x
$$

### What would you predict for the price of a 20TB drive?

```python
## A reusable linear regression function 
def price_from_capacity(capacity: float) -> float:
    return b_0 + b_1 * capacity
    
```

```python
price_20TB = price_from_capacity(20)

print(f"The linear model predicts a drive with capacity of 20TB to cost ${price_20TB:.2f}")
```

#### A 20 TB drive on Amazon.com was listed at $1023.24. According to the model, does this seem like a good buy? How much would you save compared to what you expected to pay?
We would need to calculate residual standard deviation
$$
s_e = \sqrt \frac{\Sigma e^2}{n-2}
$$

```python
## I'll need to make a sum of all errors squared
y_pred = price_from_capacity(x)
error = (y - y_pred)

standard_error = np.sqrt(np.sum(error**2) / (len(error) - 2))

print(y_pred)
print(error)
print(standard_error)
```

Then we need to find the residual of capacity(20), then find the z-score of that:

```python
## then we can compare the y value with the predicted y value as a ratio of the standard error
drive_on_amazon = {
    "size": 20,
    "price": 1023.24,
}

drive_on_amazon["predicted_price"] = price_from_capacity(drive_on_amazon["size"])
drive_on_amazon["residual_z_score"] = (drive_on_amazon["price"] - drive_on_amazon["predicted_price"]) / standard_error

print(drive_on_amazon)
print("An Amazon 20TB drive is overpriced by a z score of ", drive_on_amazon["residual_z_score"])
print(f"You would have overpaid by ${drive_on_amazon['price'] - drive_on_amazon['predicted_price']:.2f}.")
```

Based on the linear regression model, a 20TB device on Amazon, would not be a good buy. It would be overpriced, exceeding expectations by 3.15 standard deviations. 
According to the model, you could save $254.51 by buying a model closer to the regression line. 
### Does the model over or under estimate the price?
This model underestimated the price.

## Compute the residuals

Lets try making a 2d numpy array of the residuals

```python
residuals_array = np.column_stack((x, error))
```

### Which residual contributes the most to the sum that is mini-mized by the least squares criterion?

```python
largest_residual = max(residuals_array.max(), residuals_array.min(), key=abs)
print(largest_residual)
```

The answer is the residual 130.47

### Seven of the residuals are negative. What does that mean about those drives? Be specific and use the correct units.

## For the hard drive data, find and interpret the value of R2.
We found the correlation earlier and the R2 is simply squared correlation


```python
r_squared = correlation ** 2
print(r_squared)
```

This means that 95% of the variability in price is accounted for by the variation in capacity.
