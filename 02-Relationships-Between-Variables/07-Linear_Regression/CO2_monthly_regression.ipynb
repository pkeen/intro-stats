```python
import pandas as pd

df = pd.read_csv("../../data/co2monthly_2020.txt", sep="\t")
```

```python
df.head()

df["CO2"] = pd.to_numeric(df["CO2"], errors="coerce")
```

```python
# lets just go through the maths manually this time and see if we get the same as the exercise in book (70)
import numpy as np
year_frac = np.asarray(df["Year.frac"], dtype=float)
co2       = np.asarray(df["CO2"], dtype=float)

# NaN-safe means (pandas mean skips NaN by default; this matches that)
year_frac_bar = np.nanmean(year_frac)
co2_bar       = np.nanmean(co2)

# Filter out NaNs for manual sums/counts
mask_c = ~np.isnan(co2)

cv = co2[mask_c]

# Calculate
s_year_frac = np.sqrt(np.sum((year_frac - year_frac_bar)**2)/len(year_frac)-1)
s_co2 = np.sqrt(np.sum((cv - co2_bar)**2)/len(cv)-1)

print(s_co2)

print(df["CO2"].mean())
print(co2_bar)

```
