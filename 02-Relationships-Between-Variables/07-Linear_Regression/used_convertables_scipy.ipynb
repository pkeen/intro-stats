```python
import pandas as pd

df = pd.read_csv("../../data/used_convertables_2020.txt", sep="\t")
```

```python
from scipy.stats import linregress

slope, intercept, r_value, p_value, std_err = linregress(df['Age'], df['Price'])
# print(slope, intercept, r_value, p_value, std_err)
print(f"the linear regression equation is Price = {intercept:.2f} + {slope:.2f}(years)")
```

$$
\hat {Price} = 32791.12 - 984.49 \cdot years
$$

```python
import matplotlib.pyplot as plt

y_pred = intercept + slope * df['Age']
# Plot
plt.scatter(df['Age'], df['Price'], label='Data points')
plt.plot(df['Age'], y_pred, color='red', label='Regression line')
plt.title('Simple Linear Regression')
plt.xlabel('Age')
plt.ylabel('Price')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# what might a fair price be for a 10 year old convertable
predicted_10_year_price = intercept + slope * 10
print(f"The model predicts a 10 year old convertable may be worth ${predicted_10_year_price:.0f}")
```

```python
# most expensive car
# The dataset actually (really!) included another car, which we initially set aside. It is a 6-year old Mclaren MP4-12C, offered for sale for a mere $124,998. What does your linear model say about this car and its price?
price_6year_predicted = intercept + slope * 6
mclaren_true_price = 124998
mclaren_residual = mclaren_true_price - price_6year_predicted
print(f"My linear model would estimate a fair price of {price_6year_predicted:.0f}, giving a residual of {mclaren_residual:.0f}")
```
